## 1. Creating NumPy Arrays
While you can turn Python lists into arrays using `np.array()`, Machine Learning algorithms frequently require initializing arrays from scratch (e.g., creating blank matrices for weights, bias vectors set to zero, or tracking values over a linear range).

[sin(x[i]) * exp(-x[i]) for i in range(n)]

np.sin(x) * np.exp(-x)

**Note:** NumPy functions such as np.sin and np.exp are true vectorized operations. np.vectorize is mainly a convenience wrapper around a Python function; it does not automatically provide the same speedup.

### loops versus NumPy operations

The important idea is not to memorize a timing number. It is to observe that NumPy universal functions operate on the whole array, while a Python loop handles one element at a time.

````python
import time

large_x = np.linspace(0, 2 * np.pi, 200_000)

start = time.perf_counter()
# list comrehension
loop_result = [np.sin(value) * np.exp(-value) for value in large_x]
loop_result = np.array(loop_result)
loop_seconds = time.perf_counter() - start

start = time.perf_counter()
numpy_result = np.sin(large_x) * np.exp(-large_x)
numpy_seconds = time.perf_counter() - start

print('Maximum difference:', np.max(np.abs(loop_result - numpy_result)))
print('Loop seconds:', round(loop_seconds, 5))
print('NumPy seconds:', round(numpy_seconds, 5))
````

````python
import math

n = 100000
x = np.linspace(0, 2*np.pi, n)

# list comprehension
%timeit y = [math.sin(x[i]) * math.exp(-x[i]) for i in range(n)]
# use numpy ufuncs
%timeit y = np.sin(x) * np.exp(-x)
````

**What is Vectorization?**

Vectorization refers to one or both of the following, depending on the context, both of which are used in NumPy’s universal functions (ufuncs):

(1) Applying operations to all elements of a collection at once.

(2) Utilizing hardware instruction sets for parallel computation of a single instruction (SIMD).

````python
n=100000
x = np.linspace(0, 10, n)


def f(v): # apply f to one scalar value v
    return math.sin(v) * math.exp(-v)

F = np.vectorize(f)

# list comprehension
%timeit y = [f(x[i]) for i in range(n)]

# use map
%timeit y = list(map(f, x))

# use numpy's vectorize
%timeit y = F(x)

# use numpy's ufunc
%timeit y = np.sin(x) * np.exp(-x)
````

## 1. What is a Vector?

In Machine Learning, a **vector** is just a list of numbers representing the **features** of a single item or sample.

For example, a user's ratings for 3 movies can be a vector:
- User A rated: Action(5), Comedy(2), Drama(4)

In NumPy, a vector is simply a 1D array.

````python
import numpy as np

user_A = np.array([5, 2, 4])
user_B = np.array([4, 3, 5])

print("User A:", user_A)
print("User B:", user_B)
````

- **Scalar Multiplication:** Multiplying the whole vector by a single number.

## 3. Euclidean Distance: Measuring Similarity

How do we know if two users have similar tastes? We calculate the distance between their vectors in space.

The most common way is the **Euclidean Distance**. The formula is:

$$ \mathbf{A} = [ A_1, A_2, \cdots, A_n ]$$
$$ \mathbf{B} = [ B_1, B_2, \cdots, B_n ]$$

$$d = \sqrt{\sum_{i=1}^{n} (A_i - B_i)^2}$$


Similarit is almost the inverse of $$s \approx \frac{1}{d}$$


In plain English:
1. Subtract vector B from vector A
2. Square the differences
3. Sum all the squared differences
4. Take the square root of the sum

Let's implement this step-by-step in NumPy!

````python
import math

def find_similarity(v1, v2):
    # Step 1: Subtract
    diff = v1 - v2
    print("Difference:", diff)
    
    # Step 2: Square the differences
    squared_diff = diff ** 2
    print("Squared differences:", squared_diff)
    
    # Step 3: Sum them up
    sum_squared = np.sum(squared_diff)
    print("Sum of squares:", sum_squared)
    
    # Step 4: Square root
    distance = np.sqrt(sum_squared)
    print("Euclidean Distance:", distance)
    
    
    # Step 5: Invert the distance
    similarity = 1.0 / (distance + 1e-8)

    return similarity

s = find_similarity(user_A, user_B)
print(f"Similarity between user A and user B:{s}")
````

## 3. The NumPy Shortcut (`np.linalg.norm`)

Instead of writing those 4 lines every time, NumPy has a built-in function to calculate this distance (also known as the vector norm).

````python
# Calculating distance in one line!
quick_distance = np.linalg.norm(user_A - user_B)
print("Distance using linalg.norm:", quick_distance)
````

## 4. Cosine Similarity

Cosine Similarity measures how similar two vectors are by comparing the angle between them, rather than their lengths. 

More explanations in the near future. Don't worry :)

````python
dot_product = np.dot(user_A, user_B)
norm_A = np.linalg.norm(user_A)
norm_B = np.linalg.norm(user_B)

cosine_similarity = dot_product / (norm_A * norm_B)

print("Cosine Similarity:", cosine_similarity)
````

## Visual intuition: vector length and direction

A two-dimensional plot makes the geometry visible. The Euclidean distance depends on the endpoints, while cosine similarity depends on the angle between the vectors.

````python
try:
    import matplotlib.pyplot as plt
except ImportError:
    print('Matplotlib is optional for this visualization.')
else:
    vectors = {
        'a': np.array([2.0, 1.0]),
        'b': np.array([4.0, 2.0]),
        'c': np.array([-1.0, 2.0]),
    }

    colors = {
        'a': 'red',
        'b': 'blue',
        'c': 'green',
    }

    linestyles = {
        'a': '-',
        'b': '--',
        'c': ':',
    }

    origin = np.zeros(2)

    fig, ax = plt.subplots(figsize=(6, 5))

    for name, vector in vectors.items():
        print(name)

        # Draw vector shaft
        ax.plot(
            [origin[0], vector[0]],
            [origin[1], vector[1]],
            color=colors[name],
            linestyle=linestyles[name],
            linewidth=2,
            label=name,
        )

        # Draw arrowhead
        ax.annotate(
            '',
            xy=(vector[0], vector[1]),
            xytext=(vector[0] * 0.85, vector[1] * 0.85),
            arrowprops=dict(
                arrowstyle='->',
                color=colors[name],
                linewidth=2,
            ),
        )

    ax.set_xlim(-2, 5)
    ax.set_ylim(-1, 3)

    ax.axhline(0, color='gray', linewidth=0.8)
    ax.axvline(0, color='gray', linewidth=0.8)

    ax.grid(True)
    ax.legend()
    ax.set_title('Length and direction of vectors')
    ax.set_xlabel('Feature 1')
    ax.set_ylabel('Feature 2')

    plt.show()
````

## 5. Mini-Project: A Simple Recommendation System

Let's say we have a target user, and we want to find out which other user is most similar to them so we can recommend movies.

**Rule:** The *smaller* the distance, the *more similar* they are.

````python
# Ratings for [Action, Comedy, Drama, Sci-Fi]
target_user = np.array([5, 1, 1, 5])

# Database of other users
user_1 = np.array([4, 2, 1, 4])  # Loves Action and Sci-Fi
user_2 = np.array([1, 5, 5, 2])  # Loves Comedy and Drama
user_3 = np.array([5, 1, 2, 4])  # Loves Action and Sci-Fi

# Calculate distances
dist_1 = np.linalg.norm(target_user - user_1)
dist_2 = np.linalg.norm(target_user - user_2)
dist_3 = np.linalg.norm(target_user - user_3)

print(f"Distance to User 1: {dist_1:.2f}")
print(f"Distance to User 2: {dist_2:.2f}")
print(f"Distance to User 3: {dist_3:.2f}")
````

**Conclusion:** The distance to User 3 is the smallest. Therefore, `target_user` is most similar to `user_3`. If `user_3` likes a new Action movie, we should recommend it to `target_user`!

## 2. The Magic of Broadcasting

What if we want to modify our data? For example, suppose a new zoning law gives every house an extra **20 square meters** of land. The **Area** should increase by 20, while **Rooms** and **Age** remain unchanged.

Without NumPy, we would need a `for` loop to update every row one by one. NumPy provides a much smarter solution called **Broadcasting**.

Broadcasting allows NumPy to automatically perform operations between arrays of different shapes, as long as their dimensions are compatible.

Suppose our feature matrix is:

````python
X = np.array([
    [2000, 3, 10],
    [1500, 2, 5],
    [1800, 3, 8]
])
````

We create a vector that describes how each feature should change:

````python
adjustment = np.array([20, 0, 0])
````

This vector means:

- **+20** → Add 20 to the Area
- **+0** → Do not change the number of Rooms
- **+0** → Do not change the Age

### Step 1: Original Shapes

```
X shape = (3, 3)

[[2000    3   10]
 [1500    2    5]
 [1800    3    8]]

adjustment shape = (3,)

[20   0   0]
```

Notice that `X` is a **3 × 3 matrix**, while `adjustment` is only a **1D vector**.

### Step 2: Broadcasting

NumPy automatically treats the vector as if it were copied to every row.

Conceptually, it behaves like this:

```
[[20   0   0]
 [20   0   0]
 [20   0   0]]
```

**Important:** NumPy does **not** actually create this larger array in memory. It only behaves as if it exists, making broadcasting both **fast** and **memory efficient**.

### Step 3: Element-wise Addition

Now NumPy adds corresponding elements:

```
[[2000    3   10]      [[20   0   0]
 [1500    2    5]   +   [20   0   0]
 [1800    3    8]]      [20   0   0]]
```

Result:

```
[[2020    3   10]
 [1520    2    5]
 [1820    3    8]]
```

````python
# The adjustment vector: +20 to Area, +0 to Rooms, +0 to Age
adjustment = np.array([20, 0, 0])

# NumPy broadcasts the vector to every row automatically
X_adjusted = X + adjustment

print("Original Dataset:\n", X)
print("\nAdjusted Dataset:\n", X_adjusted)
````

### Why Broadcasting is Powerful

Instead of writing loops, we simply write:

```python
X + adjustment
```

NumPy automatically applies the operation to every row, making the code:

- Shorter
- Easier to read
- Much faster than Python loops

## Broadcasting Rules

NumPy can perform operations on arrays with different shapes by following these rules:

### Rule 1: Compare Shapes from Right to Left

NumPy starts comparing the dimensions from the **last (rightmost)** dimension and moves to the left.

Example:

```
(3, 4)
   (4)
```

The last dimensions are both `4`, so they are compatible.

---

### Rule 2: Dimensions are Compatible if They are Equal

If two dimensions are the same, NumPy can operate on them.

Example:

```
(5, 3)
(5, 3)
```

Every dimension matches.

✅ Broadcasting works.

---

### Rule 3: A Dimension of 1 Can Be Stretched

If one dimension is `1`, NumPy automatically repeats it to match the other array.

Example:

```
(3, 4)
(1, 4)
```

```
(2, 3, 4)
(1, 1, 4)
```

```
(2, 3, 4)
(1, 4)
```

```
(2, 3, 4)
(,4)
```

```
(1, 3, 4)
(2,1,1)
```

The first dimension `1` is stretched to `3`.

Resulting shape:

```
(3, 4)
```

---

### Rule 4: Missing Dimensions are Treated as 1

If one array has fewer dimensions, NumPy assumes missing dimensions are `1`.

Example:

```
Matrix:  (3, 4)
Vector:     (4)
```

NumPy interprets the vector as:

```
(1, 4)
```

Then broadcasts it to:

```
(3, 4)
```

---

### Rule 5: Otherwise, Broadcasting Fails

If the dimensions are neither equal nor `1`, broadcasting is impossible.

Example:

```
(3, 4)
(2, 4)
```

The first dimensions are `3` and `2`.

❌ Neither is `1`, so NumPy raises a **ValueError**.

---

## Summary

Broadcasting works when, for every dimension (from right to left):

- ✅ The dimensions are equal, **or**
- ✅ One of the dimensions is `1`

Otherwise:

- ❌ Broadcasting is not possible.

## 3. Preprocessing: Mean Centering

Many learning algorithms and optimization procedures can be sensitive when features have wildly different scales (e.g., Area is around 100, but Rooms is around 3). 

One common fix is **Mean Centering**. The formula is:
$$ X_{centered} = X - \mu $$
(where $\mu$ is the mean of each feature).

**Crucial Concept: `axis=0`**
When calculating the mean of a matrix, we must tell NumPy *how* to calculate it. 
- `axis=0` means "calculate across rows" (giving us the mean of each column/feature).

````python
# Calculate the mean of each feature (column)
feature_means = np.mean(X, axis=0)
print("Mean of [Area, Rooms, Age]:", feature_means)

# Use broadcasting to subtract the means from the entire dataset
X_centered = X - feature_means
print("\nMean-Centered Dataset:\n", X_centered)

# Notice how the values are now centered around zero!
````

## 4. Matrix Aggregations and the Power of `axis`
When working with feature matrices (Rows = Samples, Columns = Features), we frequently need to extract summary statistics like the mean, sum, or standard deviation.

- `np.mean(X)` or `X.mean()` without any arguments calculates the metric over the **entire matrix**.
- `axis=0`: Performs operations **down the columns** (collapsing rows). This gives you a statistic *per feature* (e.g., average age of all patients).
- `axis=1`: Performs operations **across the rows** (collapsing columns). This gives you a statistic *per sample* (e.g., average score across multiple tests for a single student).
- `keepdims=True`: Retains the original structural dimensions (preserves a 2D matrix shape), preventing common broadcasting dimension alignment errors.

````python
# Create a dummy feature matrix: 3 samples, 2 features (e.g., [Age, Income])
X_features = np.array([
    [25, 50000],
    [45, 85000],
    [35, 60000]
])

# 1. Overall Mean
print("Overall Matrix Mean:", np.mean(X_features))

# 2. Per-feature Mean (Mean down columns)
feature_means = np.mean(X_features, axis=0)
print("Mean of each individual feature (Age mean, Income mean):", feature_means)

# 3. Structural safety using keepdims=True
safe_feature_means = np.mean(X_features, axis=0, keepdims=True)
print('\n--------------------------\n')
print("Feature mean without keepdims:", feature_means) # (2,) Vector
print("Feature mean shape without keepdims:", feature_means.shape) # (2,) Vector
print('\n--------------------------\n')
print("Feature mean with keepdims:", safe_feature_means)   # (1, 2) Matrix
print("Feature mean shape with keepdims:", safe_feature_means.shape)   # (1, 2) Matrix
````

## Data leakage warning: scaling train and test data

The mean and standard deviation used for preprocessing should be learned from the training data only. The test set must be transformed with those training statistics.

````python
X_all = np.array([
    [2.0, 100.0],
    [3.0, 120.0],
    [4.0, 140.0],
    [5.0, 160.0],
])

X_train = X_all[:3]
X_test = X_all[3:]

train_mean = X_train.mean(axis=0)
train_std = X_train.std(axis=0)
safe_train_std = np.where(train_std == 0, 1.0, train_std)

X_train_scaled = (X_train - train_mean) / safe_train_std
X_test_scaled = (X_test - train_mean) / safe_train_std

print('Training mean:', train_mean)
print('Training-scaled data:\n', X_train_scaled.round(3))
print('Test-scaled data:\n', X_test_scaled.round(3))
````

### Safe standardization when a feature is constant

A constant feature has zero standard deviation. It carries no variation and can cause division by zero, so we use a safe denominator.

````python
constant_data = np.array([
    [1.0, 10.0],
    [2.0, 10.0],
    [3.0, 10.0],
])

mean = constant_data.mean(axis=0)
std = constant_data.std(axis=0)
safe_std = np.where(std == 0, 1.0, std)
safe_scaled = (constant_data - mean) / safe_std

print('Original std:', std)
print('Safe std:', safe_std)
print('Safe standardized data:\n', safe_scaled)
````
