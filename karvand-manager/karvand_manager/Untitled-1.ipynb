```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

np.random.seed(42)
group1 = np.random.normal(2,1, 50)
group2 = np.random.normal(8, 1, 50)
x = np.concatenate([group1, group2]).reshape(-1, 1)
y = np.array([0]*50 + [1]*50)
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x, y)
print(knn.score(x, y))

plt.scatter(x[y == 0], np.zeros(50), label = "Group1", color = "green")
plt.scatter(x[y == 1], np.zeros(50), label = "Grpup2", color = "yellow")
plt.xlabel("X")
plt.legend()
plt.show()
```

```python
from scipy import stats

girl = stats.norm(4, 1).pdf(5)
boy = stats.norm(7, 1).pdf(5)
print(round(girl, 4))
print(round(boy, 4))
if girl > boy:
    print("The girls density is greater")
else:
    print("The boys density is greater")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

np.random.seed(0)
x = np.random.rand(50, 1) * 10
noise = np.random.normal(0, 1, 50)
y = 5 + 3 * x.ravel() + noise
print(y.shape)
model = LinearRegression()
model.fit(x, y)
print(round(model.coef_[0], 4))
print(round(model.intercept_, 4))

x_line = np.linspace(0, 10, 50).reshape(-1, 1)
y_fit = model.predict(x_line)
y_true = 5 + 3 * x_line

plt.scatter(x, y, label = "data")
plt.plot(x_line, y_fit, color = "red", label = "fitted line")
plt.plot(x_line, y_true, color = "green", linestyle="--", label = "real line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("linear regression")
plt.legend()
plt.show()
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

np.random.seed(0)
group1 = np.random.normal(2, 1, 50)
group2 = np.random.normal(8, 1, 50)

x = np.concatenate([group1, group2]).reshape(-1, 1)
labels = np.array([0] * 50 + [1] * 50)
print(x.shape, labels.shape)

model = LogisticRegression()
model.fit(x, labels)
accuracy = model.score(x, labels)
print(accuracy)
```
